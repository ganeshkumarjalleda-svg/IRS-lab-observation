```python
import xml.etree.ElementTree as ET
import networkx as nx

# XML data
xml_data = """
<web>
    <page name="A">
        <link>B</link>
        <link>C</link>
    </page>

    <page name="B">
        <link>C</link>
        <link>D</link>
    </page>

    <page name="C">
        <link>A</link>
        <link>D</link>
    </page>

    <page name="D">
        <link>A</link>
    </page>
</web>
"""

# Step 1: Parse XML
root = ET.fromstring(xml_data)

# Step 2: Create web graph
G = nx.DiGraph()

for page in root.findall("page"):
    name = page.get("name")
    G.add_node(name)

    for link in page.findall("link"):
        G.add_edge(name, link.text)

# Display graph
print("Web Graph:")
print("Nodes:", list(G.nodes()))
print("Edges:", list(G.edges()))

# Step 3: Topic-specific preference
# Suppose topic is related to pages A and C
personalization = {
    "A": 0.5,
    "B": 0.0,
    "C": 0.5,
    "D": 0.0
}

# Step 4: Calculate Topic-Specific PageRank
pagerank = nx.pagerank(
    G,
    alpha=0.85,
    personalization=personalization
)

# Step 5: Display PageRank
print("\nTopic-Specific PageRank:")

for page, rank in pagerank.items():
    print(page, ":", round(rank, 4))
```
